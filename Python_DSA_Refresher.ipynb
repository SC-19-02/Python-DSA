```python
# QUESTION 1

# asks the user to input the number of days, converts the input into an integer and stores the value in a variable called days
days = int(input("Enter the number of days: "))

# calculates the number of seconds present in the days provided by the user using the formula: 24 hours * 60 minutes * 60 seconds and stores the value in a variable called seconds
seconds = days * 24 * 60 * 60

# displays the number of seconds to the user
print(f"Number of seconds is: {seconds} seconds")
```

```python
# QUESTION 2

# imports the math module which is needed for pi
import math

# asks the user to input the radius and converts the input into a floating point number and stores the value in a variable called radius
radius = float(input("Enter the radius: "))

# calculates the volume of the sphere using the formula 4/3*pi*r^3 where r is the radius provided by the user and stores the value in a variable called volume
volume = (4/3) * math.pi * (radius ** 3)

# displays the volume of the sphere to the user
print(f"Volume of sphere: {volume}")
```

```python
# QUESTION 3

# function to calculate the area of a square which has a parameter which is the side length of the square
def area(side_length):
  # area = length * width = side_length * side_length in the case of a square
  return side_length * side_length

# function to calculate the perimeter of a square which has a parameter which is the side length of the square
def perimeter(side_length):
  # perimeter = 4 * side length of the square
  return 4 * side_length

# asks the user to input the side length of the square and converts the input into a floating point number and stores the value in a variable called side_length
side_length = float(input("Enter the side length of the square: "))

# displays the area of the square to the user
print("Area:", area(side_length))

# displays the perimeter of the square to the user
print("Perimeter:", perimeter(side_length))
```

```python
# QUESTION 4

# function to check whether a character provided by the user is uppercase or lowercase which has a parameter which is the character provided by the user
def check_letter_case(char):
  if char.isupper():
    return "Character is uppercase"
  elif char.islower():
    return "Character is lowercase"
  else:
    return "Input provided is not an alphabetic character"

# asks the user to input a character and stores the value in a variable called char
char = input("Enter a character: ")

# calls the function check_letter_case and passes the character provided by the user as an argument
result = check_letter_case(char)

# displays the result to the user
print(result)
```

```python
# QUESTION 5

# initializes x to 0 and y to 20
x = 0
y = 20

# while loop that subtracts 4 from the y value and adds 2/y to the x value in each iteration and stops when y becomes less than 6
while True:
  y = y - 4
  x = x + (2/y)

  if y < 6:
    break

# displays the final value of x to the user
print(f"The final value of x is: {x}")
```

```python
# QUESTION 6

# creates an empty list/array that stores the 5 values given by the user in a variable called values
values = []

# runs a for loop 5 times, asks the user to input a value and converts the input into a floating point number each time and stores it in a variable called num
for i in range(5):
    num = float(input(f"Enter value {i+1}: "))
    # adds the value provided by the user to the values array each time
    values.append(num)

# calculates the average by dividing the sum of the values by the array length and stores the result in a variable called average
average = sum(values) / len(values)

# displays the average of the values to the user
print("The average of the values is:", average)
```
